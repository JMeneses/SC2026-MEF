**Elementos condicionais permitem criar árvores de decisão no vosso código.**

Em Python, as estruturas/árvoes de decisão permitem que seu programa escolha diferentes caminhos de execução com base em condições específicas.

Para construir condições, usam-se operadores de comparação e operadores lógicos:

**Operadores de comparação**: ```==``` (igual), ```!=``` (diferente), ```>``` (maior), ```<``` (menor), ```>=``` (maior ou igual), ```<=``` (menor ou igual)
Operadores lógicos: and (e), or (ou), not (não)





````python
# A estrutura if avalia uma condição e executa um bloco de código somente se essa condição for verdadeira.
# A estrutura else aponta um caminho alternativo no caso da condição em análise ser falsa.
idade = 29
if idade >= 18:
    print("É maior de idade, pode tirar a carta de condução da categoria B.")
else:
    print("É menor de idade, não pode tirar a carta de condução da categoria B.")

# A estrutura elif permite testar a veracidade de várias condições de forma sequencial.
if idade < 13:
  print("É uma criança.")
elif idade < 18:
  print("É um adolescente.")
elif idade < 30:
  print("É um jovem adulto.")
elif idade < 40:
  print("É um adulto.")
elif idade < 50:
  print("É um jovem sénior.")
elif idade < 60:
  print("É um sénior.")
else:
  print("É um idoso (a.k.a. velho).")

# Podem-se misturar condições com operadores lógicos
if idade >= 12 and idade <= 29:
  print("Pode tirar o cartão jovem.")
else:
  print("Já não pode tirar o cartão jovem.")
````

**Funções são uma forma prática de organizar código e torná-lo mais funcional.**



````python
# Definir uma função
def soma_de_inteiros(inteiro1, inteiro2):
  return inteiro1 + inteiro2

# Chamar uma função
print("Soma:",soma_de_inteiros(3, 12))

# Uma função não tem obrigatóriamente de retornar um valor pode manipular um valor global externo
# Importante perceber o scopus das varáveis que estamos a lidar, o espaço onde estão definidas
area = 20
def retira_area():
  global area
  area -= 5

print("Antes da função:", area)
retira_area()
print("Depois da função:", area)
````

````python
# Funções podem ser chamadas de forma recursiva.

def fatorial(n):
  if n == 0:
    return 1
  else:
    return n * fatorial(n - 1)

print("Resultado do factorial:",fatorial(6))
````

**Ciclos em Python**

As estruturas de ciclos (*loop*) permitem que se execute um bloco de código repetidamente, seja por um número definido de vezes (*for*) ou enquanto uma condição for verdadeira (*while*).

````python
# Exemplo de ciclo for
# Neste exemplo, o ciclo percorre a lista frutas e exibe cada elemento individualmente.

frutas = ["maçã", "banana", "laranja"]
for fruta in frutas:
    print(fruta)
````

````python
# A função range() permite criar sequências de números

for i in range(5):  # De 0 a 4
    print(i,end="")

print(";")
for i in range(1, 6):  # De 1 a 5
    print(i,end="")

print(";")
for i in range(0, 11, 2):  # De 0 a 10, saltando de 2 em 2
    print(i,end="")

print(";")
````

````python
# O loop while continua a executar um bloco de código enquanto uma condição for verdadeira.
# PS: Atenção! Ter sempre a certeza que a condição do while eventualmente se torna falsa, caso contrário, será um loop infinito e o programa pára a sua execução nesse loop!

contador = 0
while contador < 5:
    print(contador)
    contador += 1  # Incrementa o contador
````

````python

# Os comandos break e continue permitem controlar o fluxo dos ciclos

for i in range(10):
    if i == 5:
        break  # Sai do ciclo quando i for igual a 5
    print(i,",",end="")

print(";")

for i in range(10):
    if i % 2 == 0:
        continue  # Força a começar no i seguinte, aqui no caso de o número ser par
    print(i,",",end="")

print(";")
````

Exercício Prático: Resolver um Problema Físico

Nos Jogos Olímpicos de Inverno as pedras na modalidade de Curling seguem uma equação de velocidade linear que descreve o seu movimento em frente, dada por:

$$
v(t) = v_0 - u_k g t
$$

,onde $v_0$ é a velocidade inicial de 2.4 m/s, $g$ é a aceleração da gravidade com o valor de 9,81 m/s², e $u_k$ é o coeficiente de fricção dinâmica entre o granito e o gelo com o valor de 0.01.  

<br>

1.2.1. Utilizando o SymPy calcula qual a velocidade do objecto para o instante de 15,3 segundos?

1.2.2  Que distância total percorreu o objecto nesse tempo de 15.3 segundos?

Referências:
[The physics of sliding cylinders and curling rocks](https://doi.org/10.1119/1.1309519)

````python
t = sp.symbols('t')
v_0 = 2.4
u_k = 0.01
g = 9.8

velocity = v_0 - u_k*g*t

print("1.2.1 A velocidade do objecto para o instante de 15.3 segundos é: ", velocity.subs(t, 15.3), "m/s")

distance = sp.integrate(velocity)
travel = distance.subs(t, 15.3)

print("1.2.2 A distância total percorrida nesse mesmo tempo foi de: ",travel," m")
````

Exercício Prático: Resolver um Sistema Linear de Equações

Resolve o seguinte sistema de equações lineares usando álgebra matricial:
<br>

$$
2x + 3y = 7 \\
4x - 2y = 2 \\
$$

````python
import sympy as sp

# Define variáveis simbólicas
x, y = sp.symbols('x y')

# Define o sistema de equações
# Equação 1: 2x + 3y = 7
# Equação 2: 4x - 2y = 2

# Representa o sistema na forma matricial AS = B
# A é a matriz de coeficientes
A = sp.Matrix([
    [2, 3],
    [4, -2]
])

# B é o vetor de constantes
B = sp.Matrix([
    [7],
    [2]
])

print("Matriz de Coeficientes (A):")
sp.pprint(A)

print("\nVetor de Constantes (B):")
sp.pprint(B)

# Para resolver AS = B para S, podemos usar S = A_inv * B
# Calcula a inversa de A
A_inv = A.inv()

print("\nInversa da Matriz A (A^-1):")
sp.pprint(A_inv)

# Multiplica A_inv por B para encontrar o vetor solução S
S = A_inv * B

print("\nVetor Solução (S = A^-1 * B):")
sp.pprint(S)

print(f"\nPortanto, x = {S[0]} e y = {S[1]}. ")

# Alternativamente, SymPy fornece um método de resolução direta para matrizes
solution_direct = A.LUsolve(B)
print("\nVerificação com A.LUsolve(B):")
sp.pprint(solution_direct)
````
